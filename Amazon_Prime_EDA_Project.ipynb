```python
import pandas as pd

df = pd.read_csv('amazon_prime_titles.csv')
print("Data loaded successfully!")
display(df.head())
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
print(df['type'].value_counts())
```

```python
import matplotlib.pyplot as plt

df['type'].value_counts().plot(kind='bar', color=['skyblue', 'orange'])
plt.title('Movies vs TV Shows on Amazon Prime')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()
```

```python
import matplotlib.pyplot as plt

df['release_year'].value_counts().sort_index().plot(kind='line', figsize=(12, 6), color='green')
plt.title('Release Year Trend on Amazon Prime')
plt.xlabel('Release Year')
plt.ylabel('Count')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
df['director'].value_counts().head(10).plot(kind='bar', color='purple')
plt.title('Top 10 Directors with Most Content on Amazon Prime')
plt.xlabel('Director')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
df['country'].value_counts().head(10).plot(kind='bar', color='teal')
plt.title('Top 10 Countries with Most Content on Amazon Prime')
plt.xlabel('Country')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.show()
```
